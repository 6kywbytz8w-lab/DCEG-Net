# DCEG-Net Paper Baseline Comparison

This notebook is a lightweight entry point. The model implementations are maintained in `src/dcegnet/baselines/` rather than duplicated in a monolithic notebook.

**Prerequisite:** run the main pipeline first so that `outputs/selected_feature_list.xlsx` exists. The default run evaluates the 16 baseline models listed in the manuscript's Table 4, plus Persistence and MA7. It also imports DCEG-Net/Fusion predictions from the main results workbook when available.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src" / "dcegnet").exists():
    ROOT = ROOT.parent
if not (ROOT / "src" / "dcegnet").exists():
    raise FileNotFoundError("Open this notebook from the DCEGNet_CEEMDAN_DOI project root.")
sys.path.insert(0, str(ROOT / "src"))

from dcegnet.baselines import BaselineConfig, PAPER_BASELINES, run_baselines

print("Paper baseline count:", len(PAPER_BASELINES))
print([spec.name for spec in PAPER_BASELINES])
```

```python
config = BaselineConfig(
    data_path=ROOT / "data" / "input_dataset.xlsx",
    sheet_name="Sheet1",
    output_dir=ROOT / "outputs_baselines",
    main_feature_path=ROOT / "outputs" / "selected_feature_list.xlsx",
    main_results_path=ROOT / "outputs" / "full_results_tables_split.xlsx",
    allow_fallback_features=False,
    history=14,
    pred_horizon=1,
    seeds=(42, 52, 62),
    tune_trials=18,
    inner_cv_splits=4,
    quick_check=False,
)


metrics = run_baselines(config)
metrics.head()
```
